```
from google.colab.patches import cv2_imshow
from ultralytics import YOLO
import cv2
model = YOLO('yolov8n.pt')
image = cv2.imread('/content/Screenshot 2026-07-28 071935.png')
results = model(image)
detect = results[0].plot()
cv2_imshow(detect)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
